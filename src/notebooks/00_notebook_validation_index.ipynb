# Task 2 Notebook 검증 인덱스

이 디렉터리의 노트북은 제출자가 Python 소스 코드, fixture, Delta/DuckDB 데이터, RPC smoke path를 직접 확인한 흐름을 보여주기 위한 검증 보조 자료입니다.
루트 `README.md`와 `docs/05_validation_evidence.md`가 공식 검증 증거의 기준이며, 노트북은 사람이 데이터와 코드 흐름을 추적하기 쉽게 만든 보조 증거입니다.

## 읽는 순서

| 순서 | 노트북 | 검증 초점 | 외부 RPC 필요 여부 |
|---:|---|---|---|
| 1 | `01_rpc_provider_connection_smoke_test.ipynb` | `PipelineSettings`, `ProviderConfig`, `EthereumJsonRpcClient`가 현재 `.env`와 연결되는지 확인합니다. | 필요 |
| 2 | `02_eth_getlogs_transfer_sample_validation.ipynb` | 현재 RPC client로 `eth_getLogs`를 호출하고 ERC-20 Transfer topic/data decode 흐름을 눈으로 확인합니다. | 필요 |
| 3 | `03_fixture_etl_replay_idempotency_validation.ipynb` | fixture raw log를 Python normalizer, decoder, Delta writer에 통과시켜 재실행 멱등성을 검증합니다. | 불필요 |
| 4 | `04_accumulated_pipeline_data_freshness_validation.ipynb` | canonical Delta/DuckDB 경로의 raw, silver, gold 최신성·중복·grain을 확인합니다. | 불필요, 로컬 데이터 필요 |

## 평가자가 확인할 포인트

- 노트북이 직접 현재 Python package `cryptoquant_pipeline`를 import합니다.
- `data/delta/ethereum_logs`와 `data/analytics/ethereum_analytics.duckdb`를 canonical path로 사용합니다.
- secret, full RPC URL, API key는 출력하지 않습니다.
- 노트북 자체는 외부 RPC scheduled 수집을 검증하지 않으며, 해당 증거는 Airflow task log와 Delta/DuckDB 산출물로 분리합니다.
