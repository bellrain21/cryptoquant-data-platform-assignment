# Task 2 Notebook 검증 인덱스

이 디렉터리의 노트북은 Python 소스 코드, fixture, Delta/DuckDB 데이터, RPC smoke path를 직접 확인하실 수 있도록 정리한 검증 보조 자료입니다.
루트 `README.md`와 `docs/05_validation_evidence.md`가 공식 검증 증거의 기준이며, 노트북은 데이터와 코드 흐름을 추적과 디버깅 편의성을 목적으로 하는 자료입니다.

## 읽는 순서

| 순서 | 노트북 | 검증 초점 | 외부 RPC 필요 여부 |
|---:|---|---|---|
| 1 | `01_rpc_provider_connection_smoke_test.ipynb` | `PipelineSettings`, `ProviderConfig`, `EthereumJsonRpcClient`가 현재 `.env`와 연결되는지 확인. | 필요 |
| 2 | `02_eth_getlogs_transfer_sample_validation.ipynb` | 현재 RPC client로 `eth_getLogs`를 호출, ERC-20 Transfer topic/data decode 흐름을 눈으로 확인. | 필요 |
| 3 | `03_fixture_etl_replay_idempotency_validation.ipynb` | fixture raw log를 Python normalizer, decoder, Delta writer에 통과시켜 재실행 멱등성을 검증. | 불필요 |
| 4 | `04_accumulated_pipeline_data_freshness_validation.ipynb` | Delta/DuckDB 후보를 인벤토리화, 최신 v2 pair의 DB 추출 DataFrame, 시간대별 적재 추이, freshness, hourly gap을 확인. | 불필요, 로컬 데이터 필요 |

## 검토 시 확인하실 수 있는 포인트

- 노트북이 직접 현재 Python package `cryptoquant_pipeline`를 import합니다.
- 04번은 stale canonical default와 최신 v2 산출물을 분리해 표시합니다.
- secret, full RPC URL, API key는 출력하지 않습니다.
- 노트북 자체는 외부 RPC scheduled 수집을 검증하지 않으며, 해당 증거는 Airflow task log와 Delta/DuckDB 산출물로 분리합니다.
